```python
fig, ax = plt.subplots(figsize = (20,10), layout = 'tight')
hour_range = np.arange(0,24,0.5)
plt.plot(hour_range, yaml_usos['O3_ppbv'], label = 'Obs', linestyle = 'solid', color = 'black')
#plt.plot(hour_range, yaml_carbonbond['Conc']['O3'], label = 'CB6r5h', linestyle = 'dashed', color = 'red')
plt.plot(hour_range, yaml_geoschem['Conc']['O3'], label = 'GEOS-Chem', linestyle = 'dashed', color = 'blue')
ax.xaxis.set_minor_locator(MultipleLocator(3))
ax.xaxis.set_major_locator(MultipleLocator(6))
plt.xlabel('Hour (Mountain)')
plt.xticks(np.arange(0,24))
plt.margins(x=0)

plt.ylabel('Ozone Concentration (ppb)')
plt.grid()
plt.legend()
plt.title('Comparison of Ozone between Model and Observations \n Aug. 8, 2024')
#plt.savefig(dirpath + '/Plotting/For_F0AM_runs/initial_F0AM_GEOSChem_CB_Comparison_20240806.png', dpi =150)
plt.show()
```
