For this assignment, you will be looking at real radiosonde measurements during the [Utah Summer Ozone Study 2024](https://csl.noaa.gov/groups/csl7/measurements/2024usos/). This field campaign was held from July 15, 2024 to August 18, 2024. Your goal is to obtain the boundary layer heights throughout the campaign.

Radiosonde data includes measurements of variables such as pressure, temperature, geopotential height, dewpoint temperature, wind speed, etc. These are used to develop sounding profiles that look like this diagram below:

<img src="images/sounding_profile_20240719.jpg">

In an urban environment, it is typical to see the boundary layer height follow the sounding profiles in the image below, with the blue for nighttime, green for the morning, and red for afternoon:
<img src="images/boundary_layer_height_time_of_day.png">

We can see that the green profile looks like the profile shown in the sounding diagram of the last cell. That diagram was for a radiosonde launched at 11 AM Mountain Time. In this typical morning profile, we see an unstable layer near the surface, as the surface warms. Here, we would expect the observed lapse rate, the change in temperature over the height, $\mathbf{-\frac{dT_{atm}}{dz}}$, to be greater than the dry adiabatic lapse rate, $\Gamma= 9.8 \text{ K km}^{-1}.$ In this scenario, we can say that the top of the boundary layer is <b>at the top of this unstable layer.</b> We see an inversion above the unstable layer, which is very stable. In this region, we would expect $\mathbf{-\frac{dT_{atm}}{dz} > 0}$. This can be followed by a neutral layer that becomes more of a stable layer above, as seen in the sounding profile in the previous Markdown cell. This neutral region would be a mixed layer lower region, while the stable layer would be the mixed layer upper region.

As you go through this assignment, this will be the profile that you see the majority of the time with your 11 AM Hawthorne radiosonde data. However, the profiles may not always look so perfect with the typical profile!
